# Matrices and Nested Lists

```python
# In Python, matrices are often represented as lists of lists.
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix)
print(matrix[0])
print(matrix[1][2])  # row 2, column 3
```

### Looping Through a Matrix

```python
for row in matrix:
    for value in row:
        print(value, end=" ")
    print()
```

# Conditional Statements (if / elif / else)

```python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"

print("Grade:", grade)
```

# Classes and Objects

```python
# Classes allow you to model real-world objects.
class Student:
    def __init__(self, name, course):
        self.name = name
        self.course = course

    def introduce(self):
        return f"My name is {self.name} and I am enrolled in {self.course}"
```

```python
student1 = Student("Jessica", "AIG 230")
print(student1.introduce())
```

# Putting It All Together

```python
# TODO
""" Create a class called courses_winter 2026
with attributes course_name, code, time, grade and a method to display course info."""
class CoursesWinter2026:
    def __init__(self, course_name, code, time, grade):
        self.course_name = course_name
        self.code = code
        self.time = time
        self.grade = grade

    def display_info(self):
        return f"Course: {self.course_name} ({self.code}), Time: {self.time}, Grade: {self.grade}"
```

```python
# TODO
""" Create three instances of the courses_winter 2026 class and display their info."""
course1 = CoursesWinter2026("Artificial Intelligence for Games", "AIG 230", "Monday 10:00 AM", "A")
course2 = CoursesWinter2026("Machine Learning", "ML 301", "Wednesday 2:00 PM", "A-")
course3 = CoursesWinter2026("Deep Learning", "DL 401", "Friday 9:00 AM", "B+")

print(course1.display_info())
print(course2.display_info())
print(course3.display_info())
```
